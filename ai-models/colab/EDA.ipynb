# Khám phá dữ liệu (EDA) & Xử lý dữ liệu - Admission Prediction

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os

# Đọc dữ liệu
df = pd.read_csv('Admission_Predict.csv')
df.columns = df.columns.str.strip()
display(df.head())
print(df.info())
display(df.describe())
```

## Tạo thư mục lưu hình ảnh báo cáo

```python
os.makedirs('docs/figures', exist_ok=True)
```

## Hình 1: Phân bố biến mục tiêu (Chance of Admit)

```python
plt.figure(figsize=(8, 5))
sns.histplot(df['Chance of Admit'], kde=True, color='blue', bins=20)
plt.title('H1: Phan bo bien muc tieu (Chance of Admit)', fontsize=12, fontweight='bold')
plt.xlabel('Chance of Admit')
plt.ylabel('So luong ho so')
plt.grid(True, linestyle='--', alpha=0.5)
plt.savefig('docs/figures/fig1_target_distribution.png', bbox_inches='tight')
plt.show()
```

## Hình 2: Heatmap ma trận tương quan

```python
plt.figure(figsize=(10, 8))
corr = df.drop(columns=['Serial No.']).corr()
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt='.2f', linewidths=0.5)
plt.title('H2: Heatmap ma tran tuong quan cac dac trung', fontsize=12, fontweight='bold')
plt.savefig('docs/figures/fig2_correlation_heatmap.png', bbox_inches='tight')
plt.show()
```

## Hình 3: Scatter plot giữa CGPA và Chance of Admit

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(x='CGPA', y='Chance of Admit', data=df, hue='Research', palette='Set1', alpha=0.8)
plt.title('H3: Quan he giua CGPA va Chance of Admit', fontsize=12, fontweight='bold')
plt.xlabel('CGPA (Diem trung binh tich luy)')
plt.ylabel('Chance of Admit')
plt.grid(True, linestyle='--', alpha=0.5)
plt.savefig('docs/figures/fig3_cgpa_scatter.png', bbox_inches='tight')
plt.show()
```

## Hình 4: Boxplot kiểm tra phân bố và ngoại lai của GRE & TOEFL

```python
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
sns.boxplot(y=df['GRE Score'], color='skyblue')
plt.title('H4a: Boxplot GRE Score')

plt.subplot(1, 2, 2)
sns.boxplot(y=df['TOEFL Score'], color='lightgreen')
plt.title('H4b: Boxplot TOEFL Score')
plt.tight_layout()
plt.savefig('docs/figures/fig4_boxplots.png', bbox_inches='tight')
plt.show()
```

## Hình 5: Ảnh hưởng của kinh nghiệm Research đến Chance of Admit

```python
plt.figure(figsize=(7, 5))
research_group = df.groupby('Research')['Chance of Admit'].mean().reset_index()
research_group['Research'] = research_group['Research'].map({0: 'Khong co (0)', 1: 'Co (1)'})

sns.barplot(x='Research', y='Chance of Admit', data=research_group, palette='viridis')
plt.title('H5: Ty le trung tuyen trung binh theo Kinh nghiem Nghien cuu', fontsize=12, fontweight='bold')
plt.xlabel('Kinh nghiem nghien cuu khoa hoc')
plt.ylabel('Chance of Admit Trung binh')
plt.ylim(0, 1)
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.savefig('docs/figures/fig5_research_barplot.png', bbox_inches='tight')
plt.show()
print('Đã lưu xong toàn bộ 5 biểu đồ vào thư mục docs/figures/')
```
